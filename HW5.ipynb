```python
# ตารางความน่าจะเป็น
smoke = {True: 0.30, False: 0.70}

cancer = {
    (True, True): 0.05,
    (False, True): 0.95,
    (True, False): 0.01,
    (False, False): 0.99
}

xray = {
    ('บวก', True): 0.90,
    ('ลบ', True): 0.10,
    ('บวก', False): 0.20,
    ('ลบ', False): 0.80
}

# หาความน่าจะเป็นร่วม
def cal(s, c, t):
    return smoke[s] * cancer[(c, s)] * xray[(t, c)]


# ข้อ 41
ans41 = cal(True, True, 'บวก')
print("ข้อ 41 =", round(ans41, 4))


# ข้อ 42
ans42 = 0

for s in [True, False]:
    for c in [True, False]:
        ans42 += cal(s, c, 'บวก')

print("ข้อ 42 =", round(ans42, 4))


# ข้อ 43
cancer_positive = 0

for s in [True, False]:
    cancer_positive += cal(s, True, 'บวก')

ans43 = cancer_positive / ans42
print("ข้อ 43 =", round(ans43, 3))


# ข้อ 44
smoke_positive = 0

for c in [True, False]:
    smoke_positive += cal(True, c, 'บวก')

ans44 = smoke_positive / ans42
print("ข้อ 44 =", round(ans44, 3))


# ข้อ 45
ans45 = cancer_positive * 1000
print("ข้อ 45 =", round(ans45), "คน")
```
